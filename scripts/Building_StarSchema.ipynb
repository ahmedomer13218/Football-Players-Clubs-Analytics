```python
import pandas as pd
 
# ---------------------------------------------------------
# 0. Load
# ---------------------------------------------------------
RAW_CSV = "Data/Hubert_Football Players Stats (2024-2025)/players_data-2024_2025.csv"
df = pd.read_csv(RAW_CSV)
print(f"Loaded {df.shape[0]} rows, {df.shape[1]} columns")
pd.set_option('display.max_columns', None)
df.head()
```

```python
# ---------------------------------------------------------
# 1. Drop duplicate identity columns from merged FBref tables
#    (keep only the un-suffixed originals: Nation, Pos, Comp, Age, Born, 90s, Rk)
# ---------------------------------------------------------
dup_suffixes = [
    "_stats_shooting", "_stats_passing", "_stats_passing_types", "_stats_gca",
    "_stats_defense", "_stats_possession", "_stats_playing_time",
    "_stats_misc", "_stats_keeper", "_stats_keeper_adv"
]
identity_fields = ["Rk", "Nation", "Pos", "Comp", "Age", "Born", "90s", "MP", "Starts", "Min"]
 
cols_to_drop = [
    c for c in df.columns
    if any(c == f"{field}{suf}" for field in identity_fields for suf in dup_suffixes)
]
df = df.drop(columns=cols_to_drop)
print(f"Dropped {len(cols_to_drop)} duplicate identity columns")
 
```

```python
# ---------------------------------------------------------
# 2. Add Season constant (this file is single-season)
# ---------------------------------------------------------
df["Season"] = "2024-2025"
 
# ---------------------------------------------------------
# 3. Add per-90 calculated fields used across dashboard pages
#    (guard against divide-by-zero for players with 0 minutes)
# ---------------------------------------------------------
df["Nineties"] = df["90s"].replace(0, pd.NA)
df["Goals_per_90"] = (df["Gls"] / df["Nineties"]).round(2)
df["xG_per_90"] = (df["xG"] / df["Nineties"]).round(2)
```

```python
# ---------------------------------------------------------
# 4. DimCompetition — League x Season (this file only has one season)
# ---------------------------------------------------------
dim_competition = df[["Comp"]].drop_duplicates().rename(columns={"Comp": "League"})
dim_competition["Season"] = "2024-2025"
dim_competition.to_csv("DimCompetition.csv", index=False)
print(f"DimCompetition: {len(dim_competition)} rows")
```

```python
# ---------------------------------------------------------
# 5. DimClub — Club x League (Country not available in this file;
#    big-5 leagues map 1:1 to a country, so we derive it)
# ---------------------------------------------------------
league_country_map = {
    "eng Premier League": "England",
    "es La Liga": "Spain",
    "de Bundesliga": "Germany",
    "it Serie A": "Italy",
    "fr Ligue 1": "France",
}
 
dim_club = df[["Squad", "Comp"]].drop_duplicates().rename(
    columns={"Squad": "Club", "Comp": "League"}
)
dim_club["Country"] = dim_club["League"].map(league_country_map)
dim_club["ClubID"] = range(1, len(dim_club) + 1)
dim_club = dim_club[["ClubID", "Club", "Country", "League"]]
dim_club.to_csv("DimClub.csv", index=False)
print(f"DimClub: {len(dim_club)} rows")
if dim_club["Country"].isna().any():
    print("  NOTE: some League values didn't match the map above — check Comp naming (e.g. 'eng Premier League' vs 'Premier League')")
```

```python
# ---------------------------------------------------------
# 6. DimPlayer — one row per player
#    (single-season file, so PlayerID can be a simple surrogate)
# ---------------------------------------------------------
dim_player = df[["Player", "Nation", "Pos", "Squad", "Comp", "Age", "Born"]].drop_duplicates(subset=["Player", "Squad"])
dim_player = dim_player.merge(
    dim_club[["ClubID", "Club"]], left_on="Squad", right_on="Club", how="left"
)
dim_player["PlayerID"] = range(1, len(dim_player) + 1)
dim_player = dim_player.rename(columns={
    "Player": "PlayerName", "Nation": "Nationality", "Pos": "Position",
    "Squad": "ClubName", "Comp": "League", "Born": "BirthYear"
})
dim_player = dim_player[["PlayerID", "PlayerName", "Nationality", "Position",
                          "Age", "BirthYear", "ClubID", "ClubName", "League"]]
dim_player.to_csv("DimPlayer.csv", index=False)
print(f"DimPlayer: {len(dim_player)} rows")
```

```python
# ---------------------------------------------------------
# 7. FactPlayerStats — the metrics, joined back to PlayerID
# ---------------------------------------------------------
fact_cols = {
    "Player": "PlayerName", "Squad": "ClubName", "Season": "Season",
    "MP": "Matches", "Min": "Minutes",
    "Gls": "Goals", "Ast": "Assists", "Sh": "Shots", "SoT%": "ShotAccuracy",
    "xG": "xG", "xAG": "xA", "Goals_per_90": "GoalsPer90", "xG_per_90": "xGPer90",
    "Cmp%": "PassAccuracy", "KP": "KeyPasses", "PrgP": "ProgressivePasses",
    "TB": "ThroughBalls", "Succ": "SuccessfulDribbles",
    "SCA": "ShotCreatingActions", "GCA": "GoalCreatingActions",
    "Tkl": "Tackles", "Int": "Interceptions", "Blocks": "Blocks",
    "Clr": "Clearances", "Won": "AerialDuelsWon", "Recov": "Recoveries", "Err": "Errors",
    "Saves": "Saves", "Save%": "SavePct", "CS": "CleanSheets",
    "PSxG+/-": "GoalsPrevented", "PKsv": "PenaltiesSaved",
    "#OPA": "SweeperActions",
    "CrdY": "YellowCards", "CrdR": "RedCards",
}
missing = [c for c in fact_cols if c not in df.columns]
if missing:
    print(f"WARNING: expected columns not found, skipping: {missing}")
 
available_cols = [c for c in fact_cols if c in df.columns]
fact_player_stats = df[available_cols].rename(columns=fact_cols)
fact_player_stats = fact_player_stats.merge(
    dim_player[["PlayerID", "PlayerName", "ClubName"]],
    on=["PlayerName", "ClubName"], how="left"
)
fact_player_stats = fact_player_stats.drop(columns=["PlayerName", "ClubName"])
cols = ["PlayerID"] + [c for c in fact_player_stats.columns if c != "PlayerID"]
fact_player_stats = fact_player_stats[cols]
fact_player_stats.to_csv("FactPlayerStats.csv", index=False)
print(f"FactPlayerStats: {fact_player_stats.shape[0]} rows, {fact_player_stats.shape[1]} columns")
 
print("\nAll 4 tables written: DimPlayer.csv, DimClub.csv, DimCompetition.csv, FactPlayerStats.csv")
```

# Power BI Report

This folder is intended to contain the Power BI report file and report screenshots. The current documentation describes the intended report structure; add the actual `.pbix` file and exported screenshots before publishing the repository.

## Report Pages

The report contains the following eight pages.

### Overview

An executive-level summary showing total players, goals, assists, average age, and club distribution. Slicers support exploration by nationality, club, league, and season.

<img src="screenshots/Overview.png" alt="Overview Report" width="1000">

### PlayerProfile

A reusable profile page driven by player selection. It supports multi-position tags such as `MF,FW`, nationality flags, club badges, and drillthrough navigation.

<img src="screenshots/PlayerProfile(olmo_multi-position).png" alt="Player Profile Report" width="1000">

### Attackers

Attacking performance analysis using metrics such as goals, xG, assists, xA, shots, shot accuracy, and attacking contributions. Radar charts and scatter plots support comparisons across attacking dimensions.

<img src="screenshots/Attackers(mbappe).png" alt="Attackers Report" width="1000">

### Midfielders

Midfield analysis focused on passing, progression, and chance creation, including key passes, passing accuracy, progressive passes, through balls, xA, and creative actions.

<img src="screenshots/Midfielders(olise).png" alt="Midfielders Report" width="1000">

### Defenders

Defensive analysis covering tackles, tackles won, interceptions, blocks, clearances, aerial duels won, recoveries, and related indicators.

<img src="screenshots/Defenders(valverde).png" alt="Defenders Report" width="1000">

### Goalkeepers

Goalkeeping analysis covering saves, save percentage, clean sheets, goals conceded, post-shot expected goals, goals prevented, penalties saved, and sweeper actions.

<img src="screenshots/Goalkeepers(garcia).png" alt="Goalkeepers Report" width="1000">

### LeagueComparison

A league-level comparison page for exploring differences across the five major European leagues using available player and club indicators, comparison matrices, and supporting visuals.

<img src="screenshots/LeagueComparison.png" alt="League Comparison Report" width="1000">

### Team

A club-focused page showing results and available performance statistics, including wins, draws, losses, total games, win percentage, and home/away wins. Donut charts and attacking/defensive scatter plots support club comparisons.

<img src="screenshots/Team(barcelona).png" alt="Team Analysis Report" width="1000">

Disconnected peer tables support selected-player or selected-club highlighting in relevant scatter plots while keeping the comparison cohort visible.
